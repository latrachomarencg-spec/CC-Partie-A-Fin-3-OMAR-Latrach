## 1. Régression linéaire

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression linéaire (Linear Regression)
- **Famille :** Régression (Apprentissage supervisé)
- **Objectif :** Prédire une variable cible continue à partir d'une ou plusieurs variables explicatives en modélisant une relation linéaire entre elles.
- **Intuition et Étapes :**
  1. *Entrée :* Un ensemble de caractéristiques numériques $X$ et une cible continue $y$.
  2. *Apprentissage :* Ajustement d'une droite (ou d'un hyperplan) qui minimise la distance globale (les résidus) entre les points observés et la droite de prédiction.
  3. *Sortie :* Prédiction de valeurs continues pour de nouvelles données.
- **Formule mathématique essentielle :**
  $$\hat{y} = w^T X + b$$
  Où :
  - $\hat{y}$ est la valeur prédite (variable dépendante).
  - $X = [x_1, x_2, ..., x_n]^T$ est le vecteur des variables explicatives.
  - $w = [w_1, w_2, ..., w_n]^T$ est le vecteur des poids (paramètres du modèle appris).
  - $b$ est le biais / l'ordonnée à l'origine (paramètre appris).
  - Il n'y a pas d'hyperparamètre majeur dans une régression linéaire simple classique (pas de régularisation), contrairement aux variantes Ridge/Lasso.
- **Avantages :** Très simple à interpréter (poids explicites) et très rapide à entraîner.
- **Limites :** Sensible aux valeurs aberrantes (outliers) et incapable de capturer des relations complexes non linéaires sans transformation préalable.

### Schéma explicatif

Nous allons générer ci-dessous un schéma montrant le nuage de points, la droite de régression et la projection des résidus (les erreurs de prédiction).

```python
import numpy as np
import matplotlib.pyplot as plt

# Génération de données simples pour le schéma
np.random.seed(42)
x_schem = np.random.uniform(1, 10, 20)
y_schem = 2.5 * x_schem + 3 + np.random.normal(0, 3, 20)

# Ajustement de la droite
coef = np.polyfit(x_schem, y_schem, 1)
poly1d_fn = np.poly1d(coef)

# Tracé
plt.figure(figsize=(8, 5))
plt.scatter(x_schem, y_schem, color='blue', label='Données réelles $(x, y)$')
plt.plot(x_schem, poly1d_fn(x_schem), color='red', label='Droite de régression : $\\hat{y} = wx + b$')

# Dessin des résidus (erreurs)
for xi, yi in zip(x_schem, y_schem):
    plt.plot([xi, xi], [yi, poly1d_fn(xi)], color='gray', linestyle='--', alpha=0.6)

plt.title("Principe de la régression linéaire et des résidus")
plt.xlabel("Variable explicative (X)")
plt.ylabel("Cible (Y)")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Contextes d'utilisation
- **Cas d'usage concret (Général) :** Estimation de la demande d'énergie électrique d'un bâtiment en fonction de la température extérieure.
- **Cas d'usage concret (Métier / Finance) :** Prédiction du prix d'un bien immobilier ou estimation du budget de vente prévisionnel en fonction des dépenses marketing historiques.
- **Nature des données et prérequis :** Variables d'entrée et de sortie continues ou quantitatives. Absence de forte colinéarité entre les variables indépendantes.
- **Cas peu approprié :** Prédiction de phénomènes à croissance exponentielle ou à paliers stricts (comme des seuils d'activation biologiques) sans transformation préalable des données.
- **Métriques adaptées :**
  - L'Erreur Absolue Moyenne (MAE) : donne la grandeur moyenne des erreurs.
  - L'Erreur Quadratique Moyenne (RMSE) : pénalise plus lourdement les grandes erreurs.
  - Le Coefficient de détermination ($R^2$) : indique la proportion de la variance de la cible expliquée par le modèle.

### Analyse de l'expérience Python
Dans l'expérience réalisée :
- **Entrées :** Les variables `Xr_ent` (données d'apprentissage) et `Xr_test` (données de test) issues d'une simulation linéaire bruitée générée avec `make_regression`.

## 2. Ridge

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression Ridge (Ridge Regression / Tikhonov regularization)
- **Famille :** Régression avec régularisation (Apprentissage supervisé)
- **Objectif :** Résoudre les problèmes de surapprentissage (overfitting) et de colinéarité de la régression linéaire classique en ajoutant une pénalité L2 sur la taille des coefficients.
- **Intuition et Étapes :**
  1. *Entrée :* Caractéristiques $X$ et cible continue $y$.
  2. *Apprentissage :* Minimise la somme des carrés des résidus augmentée d'une pénalité proportionnelle au carré de la norme L2 des coefficients de poids.
  3. *Sortie :* Prédictions stabilisées où les coefficients de poids sont réduits (réduits vers zéro, mais jamais exactement nuls).
- **Formule mathématique essentielle :**
  $$\mathcal{L}_{Ridge} = \sum_{i=1}^m (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^n w_j^2$$
  Où :
  - $\alpha \ge 0$ est l'**hyperparamètre** de régularisation.
  - $w_j$ sont les poids associés aux variables explicatives (paramètres).
  - Si $\alpha = 0$, on retrouve la régression linéaire classique (MCO).
- **Avantages :** Réduit drastiquement la variance du modèle face à la colinéarité ; améliore la généralisation.
- **Limites :** Ne fait pas de sélection de variables (tous les coefficients restent non nuls) ; nécessite de normaliser les données avant l'entraînement.

### Schéma explicatif

Nous allons générer un schéma illustrant l'impact du coefficient de pénalité $\alpha$ sur la réduction des poids (le phénomène de *shrinkage*).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge

# Coefficients simulés pour illustrer l'effet du paramètre alpha
alphas = np.logspace(-2, 4, 100)
coefs = []

# Données synthétiques
np.random.seed(42)
X_toy = np.random.randn(50, 5)
y_toy = 3 * X_toy[:, 0] + 1.5 * X_toy[:, 1] + np.random.normal(0, 1, 50)

for a in alphas:
    ridge = Ridge(alpha=a)
    ridge.fit(X_toy, y_toy)
    coefs.append(ridge.coef_)

plt.figure(figsize=(8, 5))
plt.plot(alphas, coefs)
plt.xscale('log')
plt.xlabel('Hyperparamètre Alpha ($\\alpha$)')
plt.ylabel('Valeurs des coefficients (Poids $w$)')
plt.title('Effet de la régularisation Ridge (Shrinkage vers 0)')
plt.grid(True, which="both", ls="--")
plt.axvline(x=1.0, color='red', linestyle='--', label='Alpha par défaut = 1.0')
plt.legend()
plt.show()
```

### Contextes d'utilisation
- **Cas d'usage concret (Général) :** Analyse d'impacts environnementaux complexes où plusieurs facteurs météo corrélés influencent la qualité de l'air.
- **Cas d'usage concret (Métier / Finance) :** Prédiction de l'évaluation du prix d'un portefeuille financier contenant de nombreux actifs fortement corrélés entre eux (colinéarité).
- **Nature des données et prérequis :** Variables d'entrée fortement corrélées, grand nombre de caractéristiques par rapport au nombre d'échantillons ($n > m$). Centrage et réduction indispensables (via `StandardScaler`).
- **Cas peu approprié :** Lorsque l'on souhaite un modèle creux (*sparse*) pour exclure explicitement les variables non pertinentes (le Lasso sera alors bien plus adapté).
- **Métriques adaptées :** MAE, RMSE, et $R^2$ pour quantifier l'écart de prédiction sur le jeu de test.

### Analyse de l'expérience Python
Dans l'expérience réalisée :
- **Entrées :** Les données standardisées `Xr_ent` et `Xr_test`.
- **Hyperparamètres :** `alpha=1.0` (force de régularisation modérée).
- **Sortie :** Une MAE d'environ `9.37` sur le jeu de test.
- **Conclusion :** Ridge obtient une MAE quasiment identique à la régression linéaire classique (`9.35`), car le jeu de données généré comporte peu de colinéarité ou de dimensions excessives. Ridge conserve une excellente capacité de généralisation tout en stabilisant les coefficients.

```python
from sklearn.linear_model import Ridge
modele=make_pipeline(StandardScaler(),Ridge(alpha=1.0)).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 3. Lasso

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Lasso (Least Absolute Shrinkage and Selection Operator)
- **Famille :** Régression avec régularisation (Apprentissage supervisé)
- **Objectif :** Éviter le surapprentissage et effectuer automatiquement une sélection de variables en forçant certains coefficients de poids à s'annuler exactement.
- **Intuition et Étapes :**
  1. *Entrée :* Caractéristiques $X$ et cible continue $y$.
  2. *Apprentissage :* Minimise la somme des carrés des résidus augmentée d'une pénalité proportionnelle à la norme L1 absolue des coefficients.
  3. *Sortie :* Un modèle de régression parcimonieux (*sparse*) où les caractéristiques jugées superflues possèdent un coefficient égal à zéro.
- **Formule mathématique essentielle :**
  $$\mathcal{L}_{Lasso} = \sum_{i=1}^m (y_i - \hat{y}_i)^2 + \alpha \sum_{j=1}^n |w_j|$$
  Où :
  - $\alpha \ge 0$ est l'**hyperparamètre** de régularisation qui contrôle la force de la pénalité.
  - $w_j$ sont les poids associés aux caractéristiques.
- **Avantages :** Réalise une sélection automatique des caractéristiques (élimination du bruit) et produit des modèles hautement interprétables.
- **Limites :** Si plusieurs variables sont fortement corrélées, le Lasso n'en choisit arbitrairement qu'une seule ; ne fonctionne pas bien si le nombre de variables pertinentes $n$ est largement supérieur au nombre d'observations $m$.

### Schéma explicatif

Le schéma ci-dessous montre comment, à la différence de Ridge, l'augmentation du paramètre $\alpha$ du Lasso conduit à l'annulation exacte de certains coefficients (ils touchent la ligne $y=0$ et y restent).

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Lasso

# Coefficients simulés
alphas = np.logspace(-2, 2, 100)
coefs = []

# Données synthétiques
np.random.seed(42)
X_toy = np.random.randn(50, 5)
y_toy = 3 * X_toy[:, 0] + 1.5 * X_toy[:, 1] + np.random.normal(0, 1, 50)

for a in alphas:
    lasso = Lasso(alpha=a, max_iter=10000)
    lasso.fit(X_toy, y_toy)
    coefs.append(lasso.coef_)

plt.figure(figsize=(8, 5))
plt.plot(alphas, coefs)
plt.xscale('log')
plt.xlabel(r'Hyperparamètre Alpha ($\alpha$)')
plt.ylabel('Valeurs des coefficients (Poids $w$)')
plt.title('Effet de la régularisation Lasso (Sparsité des coefficients)')
plt.axhline(0, color='black', linestyle='-', alpha=0.3)
plt.grid(True, which="both", ls="--")
plt.show()
```

### Contextes d'utilisation
- **Cas d'usage concret (Général) :** Identification des biomarqueurs clés parmi des milliers de gènes mesurés pour diagnostiquer une maladie.
- **Cas d'usage concret (Métier / Finance) :** Sélection des facteurs macroéconomiques et financiers les plus influents (parmi des dizaines d'indices) pour expliquer le rendement d'un indice boursier.
- **Nature des données et prérequis :** Données tabulaires quantitatives avec potentiellement beaucoup de bruit ou de caractéristiques non informatives. Normalisation standard obligatoire.
- **Cas peu approprié :** Situation où toutes les variables explicatives ont une influence réelle, petite et répartie de façon homogène sur la cible (Ridge sera alors supérieur).
- **Métriques adaptées :** Coefficient de détermination ($R^2$), MAE et RMSE.

### Analyse de l'expérience Python
Dans l'expérience réalisée :
- **Entrées :** Données standardisées `Xr_ent` et `Xr_test`.
- **Hyperparamètres :** `alpha=0.1` et `max_iter=10000`.
- **Sortie :** Un score $R^2$ d'environ `0.986` sur le jeu de test.
- **Conclusion :** Le modèle Lasso obtient une excellente performance d'explication de la variance (98.6%), prouvant que les relations linéaires du jeu de données synthétique sont très bien capturées tout en élaguant les dimensions ou le bruit superflus.

```python
from sklearn.linear_model import Lasso
modele=make_pipeline(StandardScaler(),Lasso(alpha=.1,max_iter=10000)).fit(Xr_ent,yr_ent)
print('R²:',r2_score(yr_test,modele.predict(Xr_test)))
```

## 4. Régression polynomiale

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression polynomiale (Polynomial Regression)
- **Famille :** Régression (Apprentissage supervisé)
- **Objectif :** Modéliser des relations non linéaires complexes entre les variables explicatives et la variable cible en introduisant des puissances et des termes d'interaction des caractéristiques d'origine.
- **Intuition et Étapes :**
  1. *Entrée :* Un ensemble de caractéristiques numériques d'origine $X$.
  2. *Apprentissage / Calcul :* Les caractéristiques d'origine sont transformées de façon non linéaire en combinant leurs puissances (ex: $x^2, x^3$) et produits croisés (ex: $x_1 x_2$). Ensuite, une régression linéaire classique (ou régularisée) est ajustée sur cet espace étendu.
  3. *Sortie :* Prédictions de valeurs continues décrivant une courbe ou une surface de décision courbe.
- **Formule mathématique essentielle :**
  $$\hat{y} = w_0 + w_1 x + w_2 x^2 + w_3 x^3 + ... + w_d x^d$$
  Où :
  - $d$ est l'**hyperparamètre** définissant le degré maximum du polynôme.
  - $w_i$ sont les coefficients (poids) du modèle.
- **Avantages :** Permet de modéliser des relations fortement courbes sans changer d'algorithme de base ; s'intègre facilement via un pipeline scikit-learn.
- **Limites :** Sensible au surapprentissage (overfitting) dès que le degré $d$ augmente ; extrapolation très instable en dehors de l'intervalle des données d'entraînement.

### Schéma explicatif

Le code ci-dessous génère un schéma démontrant comment un modèle polynomial de degré 3 s'ajuste parfaitement à des données de tendance sinusoïdale, contrairement à une simple droite linéaire.

## 5. Arbre de régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Arbre de régression (Decision Tree Regressor)
- **Famille :** Régression (Apprentissage supervisé)
- **Objectif :** Prédire une variable cible continue en partitionnant l'espace des variables explicatives de manière récursive en régions homogènes.
- **Intuition et Étapes :**
  1. *Entrée :* Variables explicatives $X$ et cible continue $y$.
  2. *Apprentissage / Construction :* L'arbre découpe l'espace par des seuils successifs (splits) sur les caractéristiques pour minimiser la variance ou l'erreur quadratique au sein de chaque feuille.
  3. *Sortie :* Pour chaque nouvelle donnée, le modèle parcourt l'arbre et lui attribue la moyenne des valeurs cibles de la feuille terminale où elle atterrit.
- **Formule mathématique essentielle :**
  Le critère classique de division pour un nœud $T$ cherche à minimiser la somme des erreurs quadratiques (MSE) :
  $$\mathcal{L}(T) = \sum_{i \in T} (y_i - \bar{y}_T)^2$$
  Où :
  - $\bar{y}_T$ est la moyenne des valeurs cibles dans le nœud courant $T$.
  - Les **hyperparamètres** essentiels incluent `max_depth` (profondeur maximale de l'arbre), `min_samples_split` et `min_samples_leaf`.
- **Avantages :** Très facile à comprendre et à visualiser ; gère naturellement les variables qualitatives et quantitatives sans mise à l'échelle.
- **Limites :** Très forte tendance au surapprentissage (overfitting) si l'arbre n'est pas élagué ; modèle instable (une faible variation des données modifie complètement la structure).

### Schéma explicatif

Le code ci-dessous illustre le partitionnement caractéristique en paliers (escalier) généré par un arbre de régression sur des données courbes.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

np.random.seed(42)
x_plot = np.sort(np.random.uniform(-3, 3, 40))
y_plot = np.sin(x_plot) + np.random.normal(0, 0.15, 40)

# Ajustement de deux arbres de profondeurs différentes
tree_depth2 = DecisionTreeRegressor(max_depth=2, random_state=42).fit(x_plot.reshape(-1, 1), y_plot)
tree_depth5 = DecisionTreeRegressor(max_depth=5, random_state=42).fit(x_plot.reshape(-1, 1), y_plot)

x_grid = np.linspace(-3, 3, 200).reshape(-1, 1)

plt.figure(figsize=(8, 5))
plt.scatter(x_plot, y_plot, color='blue', label='Données réelles')
plt.step(x_grid, tree_depth2.predict(x_grid), color='red', linewidth=2, label='Profondeur 2 (Sous-apprentissage)')
plt.step(x_grid, tree_depth5.predict(x_grid), color='green', linewidth=2, label='Profondeur 5 (Bon ajustement)')
plt.title("Modélisation non-linéaire par paliers (Arbre de régression)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

- **Cas d'usage concret (Métier / Finance) :** Établissement de règles de tarification d'assurances en fonction du profil de risque du client (âge, lieu de résidence, historique d'accidents).
- **Nature des données et prérequis :** Variables d'entrée mixtes (numériques ou catégorielles). Ne requiert aucune transformation des données (ni normalisation, ni traitement de la colinéarité).
- **Cas peu approprié :** Problèmes d'extrapolations où la tendance doit se prolonger de façon continue et monotone hors de l'espace d'entraînement.
- **Métriques adaptées :** MAE, RMSE, et $R^2$.

## 6. Forêt aléatoire — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Forêt aléatoire pour la régression (Random Forest Regressor)
- **Famille :** Régression (Ensemble Learning / Bagging - Apprentissage supervisé)
- **Objectif :** Améliorer la précision de prédiction et contrôler le surapprentissage d'un arbre de décision unique en entraînant une multitude d'arbres indépendants sur des sous-ensembles aléatoires de données et en moyennant leurs prédictions.
- **Intuition et Étapes :**
  1. *Entrée :* Variables explicatives $X$ et cible continue $y$.
  2. *Échantillonnage (Bootstrap) :* Création de $N$ sous-ensembles de données par tirage aléatoire avec remise.
  3. *Apprentissage (Bagging) :* Entraînement d'un arbre de régression sur chaque échantillon. À chaque division d'un nœud, seul un sous-ensemble aléatoire de caractéristiques est candidat au fractionnement.
  4. *Agrégation :* La prédiction finale est la moyenne arithmétique des prédictions de tous les arbres individuels.
- **Formule mathématique essentielle :**
  $$\hat{y} = \frac{1}{B} \sum_{b=1}^{B} T_b(x)$$
  Où :
  - $B$ est l'**hyperparamètre** définissant le nombre d'arbres (`n_estimators`).
  - $T_b(x)$ est la prédiction issue du $b$-ième arbre de régression.
  - Les autres hyperparamètres clés sont `max_features` et `max_depth`.
- **Avantages :** Très robuste au surapprentissage grâce à la double randomisation (lignes et colonnes) ; gère efficacement les relations non linéaires complexes.
- **Limites :** Modèle volumineux en mémoire et agissant comme une "boîte noire" difficile à interpréter individuellement ; incapable d'extrapoler au-delà des bornes des données d'entraînement.

### Schéma explicatif

Le code ci-dessous illustre comment une forêt aléatoire lisse la fonction de prédiction par paliers (en vert) par rapport à un unique arbre sur-ajusté (en rouge) sur des données sinusoïdales.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

np.random.seed(42)
x_rf = np.sort(np.random.uniform(-3, 3, 50))
y_rf = np.sin(x_rf) + np.random.normal(0, 0.2, 50)

tree = DecisionTreeRegressor(max_depth=5, random_state=42).fit(x_rf.reshape(-1, 1), y_rf)
rf = RandomForestRegressor(n_estimators=50, max_depth=5, random_state=42).fit(x_rf.reshape(-1, 1), y_rf)

x_grid = np.linspace(-3, 3, 300).reshape(-1, 1)

plt.figure(figsize=(8, 5))
plt.scatter(x_rf, y_rf, color='blue', label='Données réelles')
plt.step(x_grid, tree.predict(x_grid), color='red', alpha=0.6, label='Arbre unique (Instable)')
plt.plot(x_grid, rf.predict(x_grid), color='green', linewidth=2, label='Forêt aléatoire (Lissée / Robuste)')
plt.title("Stabilisation des prédictions : Arbre vs Forêt Aléatoire")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Contextes d'utilisation
- **Cas d'usage concret (Général) :** Prédiction de la température de surface ou de la consommation d'énergie d'une ville entière.
- **Cas d'usage concret (Métier / Finance) :** Estimation de la valeur vénale d'actifs immobiliers complexes à partir de nombreuses caractéristiques (surface, géolocalisation, historique de ventes du quartier, taux d'intérêt).
- **Nature des données et prérequis :** Données tabulaires d'entrées hétérogènes (numériques et qualitatives) avec potentiellement des valeurs manquantes. Aucun besoin de mise à l'échelle.
- **Cas peu approprié :** Problèmes de séries temporelles avec une forte tendance linéaire (extrapolation) où les valeurs futures dépassent largement l'intervalle d'apprentissage.
- **Métriques adaptées :** MAE, RMSE, et $R^2$.

Dans l'expérience réalisée :

- **Hyperparamètres :** `n_estimators=80` et `random_state=42`.
- **Sortie :** Une MAE de `24.37` sur le jeu de test.
- **Conclusion :** La MAE (24.37) de la forêt aléatoire est largement inférieure à celle d'un arbre de décision unique (37.72). Cela prouve que l'agrégation de 80 estimateurs indépendants permet de réduire significativement la variance globale et d'obtenir des prédictions beaucoup plus précises et stables, même si la structure purement linéaire favorise encore la régression linéaire classique.

```python
from sklearn.ensemble import RandomForestRegressor
modele=RandomForestRegressor(n_estimators=80,random_state=GRAINE).fit(Xr_ent,yr_ent)
print('MAE:',mean_absolute_error(yr_test,modele.predict(Xr_test)))
```

## 7. Gradient Boosting — régression

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Gradient Boosting pour la régression (Gradient Boosting Regressor / HistGradientBoostingRegressor)
- **Famille :** Régression (Ensemble Learning / Boosting - Apprentissage supervisé)
- **Objectif :** Prédire une variable cible continue en combinant séquentiellement plusieurs modèles faibles (généralement des arbres de décision peu profonds), chacun corrigeant les erreurs de prédiction (résidus) de ses prédécesseurs.
- **Intuition et Étapes :**
  1. *Initialisation :* On démarre avec une prédiction constante simple (la moyenne de la cible).
  2. *Séquence de correction :* À chaque étape $t$, on calcule l'écart (le pseudo-résidu, qui correspond au gradient négatif de la fonction de perte) entre la valeur réelle et la prédiction actuelle.
  3. *Modèle faible :* On entraîne un nouvel arbre de décision pour prédire ces résidus.
  4. *Mise à jour :* On ajoute la prédiction de ce nouvel arbre au modèle global, multipliée par un taux d'apprentissage (learning rate) pour l'intégrer progressivement.
- **Formule mathématique essentielle :**
  Le modèle à l'étape $m$ s'écrit :
  $$F_m(x) = F_{m-1}(x) +
u
ho_m h_m(x)$$
  Où :
  - $F_{m-1}(x)$ est la prédiction accumulée à l'étape précédente.
  - $h_m(x)$ est l'arbre de décision faible ajusté sur les pseudo-résidus.
  - $
u \in ]0, 1]$ est l'**hyperparamètre** du taux d'apprentissage (`learning_rate`).
  - $
ho_m$ est le pas optimal trouvé par minimisation.
  - Les autres hyperparamètres clés sont `max_iter` (nombre d'itérations / d'arbres) et `max_depth`.
- **Avantages :** Précision prédictive extrêmement élevée et gestion performante des relations non linéaires complexes.
- **Limites :** Sensible au surapprentissage si le nombre d'arbres est trop grand ou si le taux d'apprentissage n'est pas bien ajusté ; processus d'entraînement séquentiel plus lent à paralléliser que les forêts aléatoires.

- **Cas peu approprié :** Jeux de données contenant un bruit extrême sans signal clair (le boosting aura tendance à surapprendre sur le bruit) ou configurations nécessitant une interprétabilité simple et immédiate des variables.
- **Métriques adaptées :** MAE, RMSE, et $R^2$.

## 8. Régression logistique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Régression logistique (Logistic Regression)
- **Famille :** Classification (Apprentissage supervisé)
- **Objectif :** Prédire la probabilité qu'une observation appartienne à une classe binaire (0 ou 1) ou multinomiale en modélisant la relation entre les variables explicatives et la cible à l'aide d'une fonction logistique.
- **Intuition et Étapes :**
  1. *Entrée :* Caractéristiques quantitatives $X$ et étiquettes de classe discrètes $y$ (ex: 0 ou 1).
  2. *Transformation sigmoïde :* Calcule une combinaison linéaire des entrées, puis lui applique la fonction sigmoïde pour projeter la valeur brute sur l'intervalle de probabilité $[0, 1]$.
  3. *Apprentissage :* Recherche les paramètres de poids en maximisant la vraisemblance (ou en minimisant la perte d'entropie croisée binary cross-entropy).
  4. *Prédiction :* Assigne la classe 1 si la probabilité calculée dépasse un seuil de décision (généralement $0.5$), sinon la classe 0.
- **Formule mathématique essentielle :**
  La probabilité prédite s'écrit :
  $$P(y=1|X) = ̂{\sigma}(w^T X + b) = \frac{1}{1 + e^{-(w^T X + b)}}$$
  Où :
  - $w$ est le vecteur des poids et $b$ le biais (paramètres appris).
  - l'**hyperparamètre** principal est la force de régularisation $C$ (inverse de la pénalité L1 ou L2).
- **Avantages :** Donne des probabilités bien étalonnées et s'interprète très facilement via les *odds ratios* (rapports des chances).
- **Limites :** Suppose une frontière de décision linéaire entre les classes ; performe mal face à des relations hautement non linéaires sans ingénierie de caractéristiques complexe.

### Schéma explicatif

Le code ci-dessous génère un schéma de la fonction sigmoïde caractéristique de la régression logistique, illustrant comment les scores linéaires continus sont écrasés sur un intervalle de probabilité $[0,1]$ avec un seuil de décision à $0.5$.

### Contextes d'utilisation

- **Cas d'usage concret (Métier / Finance) :** Évaluation du risque de crédit (*credit scoring*) pour décider de l'octroi d'un prêt bancaire, ou détection de la résiliation d'un abonnement client (attrition / churn) en télécoms.

- **Cas peu approprié :** Problèmes d'analyse d'images complexes ou de textes massifs bruts sans représentation sémantique dense, car les frontières de classification y sont intrinsèquement non linéaires.
- **Métriques adaptées :** F1-score, Précision, Rappel et ROC-AUC.

## 9. KNN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** K-plus proches voisins (K-Nearest Neighbors / KNN)
- **Famille :** Classification / Régression (Apprentissage supervisé non paramétrique)
- **Objectif :** Classer une nouvelle observation en lui attribuant la classe majoritaire parmi ses $k$ plus proches voisins dans l'espace des caractéristiques.
- **Intuition et Étapes :**
  1. *Entrée :* Données d'entraînement stockées en mémoire (caractéristiques $X$ et étiquettes $y$) et une nouvelle observation à classer.
  2. *Calcul de distance :* Calcule la distance (souvent euclidienne) entre la nouvelle observation et l'ensemble des points d'entraînement.
  3. *Sélection :* Identifie les $k$ observations de l'entraînement les plus proches.
  4. *Vote majoritaire :* Détermine la classe la plus fréquente parmi ces $k$ voisins et l'attribue en sortie.
- **Formule mathématique essentielle :**
  La fonction de prédiction pour une classe $c$ est donnée par le vote majoritaire :
  $$\hat{y} = \arg\max_c \sum_{i \in N_k(x)} I(y_i = c)$$
  Où :
  - $N_k(x)$ représente l'ensemble des $k$ voisins les plus proches de l'observation $x$.
  - $I(\cdot)$ est la fonction indicatrice qui vaut 1 si la condition est vraie, 0 sinon.
  - Le nombre de voisins $k$ (`n_neighbors`) est l'**hyperparamètre** principal du modèle.
- **Avantages :** Très simple à comprendre, n'émet aucune hypothèse sur la distribution des données (non paramétrique) et s'adapte naturellement aux problèmes multi-classes.
- **Limites :** Coûteux en mémoire et en temps lors de la phase de test (car il faut calculer toutes les distances) ; extrêmement sensible au fléau de la dimension (*curse of dimensionality*) et nécessite une mise à l'échelle des variables.

## 10. Naive Bayes

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Classifieur Naïf Bayésien (Naive Bayes Classifier / Gaussian Naive Bayes)
- **Famille :** Classification (Apprentissage supervisé probabiliste)
- **Objectif :** Classer une observation dans l'une des classes prédéfinies en calculant la probabilité conditionnelle *a posteriori* de chaque classe sachant les caractéristiques observées, et en choisissant la classe avec la probabilité la plus élevée.
- **Intuition et Étapes :**
  1. *Entrée :* Caractéristiques $X = (x_1, ..., x_n)$ et étiquettes de classe discrètes.
  2. *Hypothèse "Naïve" :* L'algorithme suppose que toutes les caractéristiques sont indépendantes les unes des autres, conditionnellement à la classe.
  3. *Calcul :* Pour chaque classe, on multiplie la probabilité *a priori* de la classe par le produit des probabilités d'observer chaque caractéristique dans cette classe (vraisemblance).
  4. *Sortie :* Classe maximisant cette probabilité finale.
- **Formule mathématique essentielle :**
  D'après le théorème de Bayes et l'hypothèse d'indépendance conditionnelle :
  $$P(y = c | x_1, …, x_n) \propto P(y = c) \prod_{j=1}^{n} P(x_j | y = c)$$
  Pour le cas continu (Gaussian Naive Bayes), la vraisemblance est modélisée par une loi normale (gaussienne) :
  $$P(x_j | y = c) = \frac{1}{\sqrt{2\pi\sigma_{jc}^2}} \exp\left( -\frac{(x_j - \mu_{jc})^2}{2\sigma_{jc}^2} \right)$$
  Où :
  - $\mu_{jc}$ et $\sigma_{jc}^2$ sont respectivement la moyenne et la variance de la caractéristique $j$ pour la classe $c$ (paramètres appris).
  - Il n'y a pas d'hyperparamètre majeur dans sa version classique standard (pas de régularisation), en dehors du lissage de Laplace pour d'autres variantes (Multinomial).
- **Avantages :** Extrêmement rapide à entraîner et à prédire ; performe étonnamment bien sur des données de grande dimension (comme le traitement du texte) même quand l'hypothèse d'indépendance est violée.
- **Limites :** L'hypothèse d'indépendance stricte entre les variables est rarement vérifiée dans la réalité ; si une catégorie d'une variable n'a pas été observée à l'entraînement pour une classe, la probabilité finale tombe à zéro (problème de fréquence nulle).

### Schéma explicatif

Le code ci-dessous génère un schéma illustrant la modélisation sous-jacente d'un Naive Bayes Gaussien pour une seule caractéristique $X$. Il représente les distributions de probabilités gaussiennes apprises pour deux classes distinctes, et montre comment la probabilité *a posteriori* détermine l'intersection qui sépare les zones de décision.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Paramètres simulés pour deux classes
mu_A, sigma_A = 2.0, 0.8
mu_B, sigma_B = 4.5, 1.2

x_range = np.linspace(-1, 8, 500)

plt.figure(figsize=(8, 4.5))
plt.plot(x_range, norm.pdf(x_range, mu_A, sigma_A), color='red', linewidth=2, label='Vraisemblance $P(X|Classe A)$')
plt.plot(x_range, norm.pdf(x_range, mu_B, sigma_B), color='blue', linewidth=2, label='Vraisemblance $P(X|Classe B)$')

# Zone d'intersection pour la frontière de décision
frontiere = 3.12 # Point d'intersection théorique approximatif
plt.axvline(frontiere, color='black', linestyle='--', label=f'Frontière de décision : X ~ {frontiere}')
plt.fill_between(x_range, 0, norm.pdf(x_range, mu_A, sigma_A), where=(x_range < frontiere), color='red', alpha=0.1)
plt.fill_between(x_range, 0, norm.pdf(x_range, mu_B, sigma_B), where=(x_range >= frontiere), color='blue', alpha=0.1)

plt.title("Principe du Naive Bayes Gaussien (Modélisation par caractéristique)")
plt.xlabel("Valeur de la caractéristique (X)")
plt.ylabel("Densité de probabilité")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

- **Cas d'usage concret (Métier / Marketing) :** Analyse de sentiments dans les avis clients (classification textuelle rapide en marketing pour trier les retours positifs, neutres ou négatifs).
- **Nature des données et prérequis :** Variables d'entrée qualitatives (Naive Bayes Multinomial) ou quantitatives continues (Naive Bayes Gaussien). Nécessite l'absence de corrélation forte entre caractéristiques pour ne pas fausser l'estimation des probabilités.
- **Cas peu approprié :** Prédiction financière où l'interaction et la dépendance entre variables (ex: taux d'intérêt, inflation, cours des actions) sont hautement critiques pour le modèle.
- **Métriques adaptées :** F1-score, Précision, Rappel et exactitude (Accuracy).

## 11. Arbre de décision

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Arbre de décision pour la classification (Decision Tree Classifier)
- **Famille :** Classification (Apprentissage supervisé)
- **Objectif :** Classer des observations en construisant une hiérarchie de règles de décision (sous forme d'arbre) basées sur les valeurs des caractéristiques d'entrée.
- **Intuition et Étapes :**
  1. *Entrée :* Caractéristiques $X$ et étiquettes de classes discrètes $y$.
  2. *Partitionnement :* L'algorithme sélectionne à chaque nœud la caractéristique et le seuil qui séparent le mieux les classes de manière à maximiser l'homogénéité (pureté) des sous-groupes enfants.
  3. *Prédiction :* Un nouvel échantillon descend le long de l'arbre en fonction des tests successifs jusqu'à atteindre une feuille, qui lui attribue sa classe majoritaire.
- **Formule mathématique essentielle :**
  La division optimale au niveau d'un nœud est déterminée en minimisant l'**Impureté de Gini** ($I_G$) ou l'**Entropie** ($H$) :
  $$I_G(T) = 1 - \sum_{i=1}^{C} p_i^2$$
  Où :
  - $C$ est le nombre total de classes.
  - $p_i$ est la proportion (probabilité) d'observations appartenant à la classe $i$ dans le nœud $T$.
  - Les principaux **hyperparamètres** sont `max_depth` (profondeur limite de l'arbre), `min_samples_split` (nombre minimal d'échantillons requis pour scinder un nœud) et `criterion` ('gini' ou 'entropy').
- **Avantages :** Extrêmement simple à interpréter et à visualiser ; ne requiert aucun prétraitement des données (comme la normalisation) et gère nativement les données manquantes ou mixtes.
- **Limites :** Très forte instabilité (une petite modification des données d'entraînement peut restructurer complètement l'arbre) ; propension naturelle au surapprentissage (overfitting) si l'arbre n'est pas restreint.

## 12. Forêt aléatoire — classification

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Forêt aléatoire pour la classification (Random Forest Classifier)
- **Famille :** Classification (Ensemble Learning / Bagging - Apprentissage supervisé)
- **Objectif :** Classer des observations dans des catégories discrètes en combinant les prédictions d'un grand nombre d'arbres de décision indépendants pour réduire la variance et éviter le surapprentissage.
- **Intuition et Étapes :**
  1. *Entrée :* Variables explicatives $X$ et étiquettes de classes discrètes $y$.
  2. *Bootstrap (Échantillonnage) :* Sélection aléatoire avec remise de sous-ensembles d'observations pour entraîner chaque arbre.
  3. *Random Subspace (Sous-espace aléatoire) :* À chaque division de nœud de chaque arbre, sélection aléatoire d'un sous-ensemble de caractéristiques pour déterminer le meilleur découpage.
  4. *Vote majoritaire :* Chaque arbre de la forêt émet une prédiction de classe. La classe finale attribuée est celle ayant obtenu la majorité absolue ou relative des votes.
- **Formule mathématique essentielle :**
  La probabilité prédite pour la classe $c$ par la forêt aléatoire est la moyenne des probabilités de chaque arbre :
  $$P(y = c | X) = \frac{1}{B} \sum_{b=1}^{B} P_b(y = c | X)$$
  Où :
  - $B$ est l'**hyperparamètre** du nombre d'arbres (`n_estimators`).
  - $P_b(y = c | X)$ est la probabilité affectée à la classe $c$ par le $b$-ième arbre.
  - Les autres hyperparamètres critiques sont `max_features` (nombre de variables à évaluer à chaque split) et `max_depth`.
- **Avantages :** Exceptionnellement robuste au surapprentissage grâce à la double injection d'aléa ; fournit d'excellents scores d'importance des variables.
- **Limites :** Modèle volumineux, lourd en mémoire et plus lent à exécuter en temps réel qu'un simple arbre ; interprétabilité directe difficile (effet "boîte noire" de l'agrégation).

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Machine à Vecteurs de Support / Séparateurs à Vaste Marge (Support Vector Machine - SVM)
- **Famille :** Classification / Régression (Apprentissage supervisé)
- **Objectif :** Trouver l'hyperplan de séparation optimal entre différentes classes en maximisant la marge (la distance géométrique) entre cet hyperplan et les points d'entraînement les plus proches.
- **Intuition et Étapes :**
  1. *Entrée :* Un ensemble de caractéristiques numériques $X$ et des étiquettes de classes discrètes $y$ (généralement -1 ou 1).
  2. *Transformation (Astuce du noyau) :* Si les données ne sont pas linéairement séparables, elles sont projetées dans un espace de dimension supérieure (via un noyau comme le RBF) où elles le deviennent.
  3. *Optimisation :* Recherche de l'hyperplan qui sépare les classes tout en maximisant la distance (marge) avec les points les plus proches (vecteurs de support).
  4. *Sortie :* Une frontière de décision géométrique robuste pour classer les nouvelles observations.
- **Formule mathématique essentielle :**
  Pour un classifieur à marge souple, le problème d'optimisation quadratique s'écrit :
  $$\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^m \xi_i \quad \text{sous la contrainte} \quad y_i(w^T \phi(X_i) + b) \ge 1 - \xi_i, \quad \xi_i \ge 0$$
  Où :
  - $w$ est le vecteur de poids définissant l'orientation de l'hyperplan et $b$ est le biais (paramètres du modèle appris).
  - $\xi_i$ sont les variables d'écart (*slack variables*) mesurant le degré de mauvaise classification de l'échantillon $i$.
  - $C > 0$ est l'**hyperparamètre** de pénalité contrôlant le compromis entre la maximisation de la marge et la tolérance aux erreurs d'entraînement.
  - $\phi(\cdot)$ représente la fonction de projection dans l'espace de Hilbert associée au noyau choisi (ex: noyau RBF contrôlé par l'hyperparamètre $\gamma$).
- **Avantages :** Très efficace dans les espaces de grande dimension (même quand $n > m$) et robuste face au surapprentissage grâce à la maximisation géométrique de la marge.
- **Limites :** Extrêmement sensible à la mise à l'échelle des données (nécessite obligatoirement un `StandardScaler`) et très coûteux en temps de calcul sur de grands ensembles de données ($O(m^3)$).

### Schéma explicatif

Le schéma ci-dessus (généré dynamiquement par le script Python) illustre le principe fondamental du SVM linéaire :
- **Ligne continue noire :** Représente l'hyperplan de décision central d'équation $w^T X + b = 0$.
- **Lignes pointillées noires :** Matérialisent les frontières de la marge à $w^T X + b = \pm 1$. La distance séparant ces deux pointillés est la marge maximale que l'algorithme cherche à maximiser géométriquement ($2 / \|w\|$).
- **Cercles verts épais :** Mettent en valeur les **vecteurs de support**. Ce sont les uniques points de données d'entraînement qui touchent ou franchissent la marge et qui déterminent à eux seuls la position finale de l'hyperplan de décision. Tous les autres points du nuage n'influent pas sur la frontière construite.

### Présentation
- **Nom :** Machine à Vecteurs de Support / Séparateurs à Vaste Marge (Support Vector Machine - SVM)
- **Famille :** Classification / Régression (Apprentissage supervisé)
- **Objectif :** Trouver l'hyperplan de séparation optimal entre différentes classes en maximisant la marge (la distance) entre cet hyperplan et les points de données les plus proches, appelés vecteurs de support.
- **Intuition et Étapes :**
  1. *Entrée :* Caractéristiques $X$ et étiquettes de classe $y$ (généralement -1 ou 1).
  2. *Optimisation :* Recherche d'un hyperplan de décision linéaire ou projection des données dans un espace de dimension supérieure (astuce du noyau / *kernel trick*) pour rendre les classes linéairement séparables.
  3. *Calcul de marge :* Maximisation de la marge géométrique tout en gérant d'éventuelles erreurs contrôlées par une variable de pénalité.
  4. *Sortie :* Frontière de décision nette et robuste maximisant la marge.
- **Formule mathématique essentielle :**
  Pour un modèle à marge souple linéaire, le problème d'optimisation s'écrit :
  $$\min_{w, b, \xi} \frac{1}{2} \|w\|^2 + C \sum_{i=1}^m \xi_i \quad \text{sujet à} \quad y_i (w^T X_i + b) \ge 1 - \xi_i, \quad \xi_i \ge 0$$
  Où :
  - $w$ est le vecteur de poids définissant l'orientation de l'hyperplan (paramètre).
  - $b$ est le biais (paramètre).
  - $\xi_i$ sont les variables d'écart (*slack variables*) mesurant les erreurs d'alignement.
  - $C > 0$ est l'**hyperparamètre** qui équilibre la maximisation de la marge et la tolérance aux erreurs de classification.
  - En utilisant un noyau Gaussien (RBF), un autre hyperparamètre clé est $\gamma$ (`gamma`), définissant le rayon d'influence d'un seul échantillon d'entraînement.
- **Avantages :** Très performant dans les espaces de grande dimension (même quand $n > m$) ; résiste bien au surapprentissage grâce à la maximisation intrinsèque de la marge.
- **Limites :** Sensible au bruit et aux valeurs aberrantes ; nécessite impérativement une mise à l'échelle rigoureuse des données ; lourd à entraîner sur de très grands ensembles de données.

### Présentation
- **Nom :** Gradient Boosting pour la classification (Gradient Boosting for Classification / HistGradientBoostingClassifier)
- **Famille :** Classification (Ensemble Learning / Boosting - Apprentissage supervisé)
- **Objectif :** Prédire l'appartenance d'une observation à une classe discrète en combinant séquentiellement plusieurs arbres de décision faibles, chacun ajusté pour corriger les erreurs de classification de l'ensemble des arbres précédents.
- **Intuition et Étapes :**
  1. *Initialisation :* Démarrage avec une prédiction constante (par exemple, le logarithme du rapport des chances de la classe majoritaire).
  2. *Séquence adaptative :* À chaque étape $t$, on calcule le gradient négatif de la fonction de perte (l'entropie croisée), représentant l'écart résiduel probabiliste pour chaque échantillon.
  3. *Ajustement :* Un nouvel arbre de décision faible est entraîné pour prédire ces résidus.
  4. *Mise à jour :* Les prédictions du nouvel arbre sont intégrées au score global via un taux d'apprentissage amortissant.
- **Formule mathématique essentielle :**
  La probabilité de classe prédite à l'aide de la fonction logistique d'un score cumulé $F_M(x)$ est :
  $$P(y=1|x) = \frac{1}{1 + e^{-F_M(x)}} \quad \text{avec} \quad F_M(x) = F_0(x) + \nu \sum_{m=1}^{M} h_m(x)$$
  Où :
  - $F_0(x)$ est la valeur d'initialisation.
  - $h_m(x)$ est l'arbre faible entraîné sur les résidus à l'étape $m$.
  - $\nu \in ]0, 1]$ est l'**hyperparamètre** du taux d'apprentissage (`learning_rate`).
  - $M$ est l'**hyperparamètre** limitant le nombre d'itérations (`max_iter`).
- **Avantages :** Précision prédictive hors pair sur données tabulaires ; gère nativement les valeurs manquantes sans imputation complexe.
- **Limites :** Processus d'apprentissage séquentiel difficile à paralléliser ; forte sensibilité aux bruits de classe si les hyperparamètres ne sont pas bien régularisés.

### Présentation
- **Nom :** AdaBoost (Adaptive Boosting / AdaBoostClassifier)
- **Famille :** Classification (Ensemble Learning / Boosting - Apprentissage supervisé)
- **Objectif :** Combiner séquentiellement une série de classifieurs très simples (généralement des arbres de décision à un seul niveau, appelés *decision stumps*) en ajustant de manière itérative le poids des observations pour focaliser l'apprentissage sur les exemples précédemment mal classés.
- **Intuition et Étapes :**
  1. *Initialisation :* Assigner un poids uniforme $w_i = 1/m$ à chaque échantillon.
  2. *Boucle séquentielle :* À chaque itération $t$, un classifieur faible est ajusté sur les données pondérées.
  3. *Calcul de l'erreur :* On calcule l'erreur pondérée du modèle et on en déduit son importance $\alpha_t$ (plus l'erreur est faible, plus son poids $\alpha_t$ est grand).
  4. *Mise à jour des poids :* On augmente le poids des données mal classées pour forcer le prochain classifieur à se concentrer sur elles, puis on normalise les poids.
  5. *Prédiction finale :* Résulte d'un vote majoritaire pondéré par les coefficients $\alpha_t$.
- **Formule mathématique essentielle :**
  Le coefficient de confiance d'un estimateur faible $h_t(x)$ est défini par :
  $$\alpha_t = \frac{1}{2} \ln\left( \frac{1 - \epsilon_t}{\epsilon_t} \right) \quad \text{avec} \quad \epsilon_t = \sum_{i=1}^m w_i^{(t)} I(y_i \neq h_t(X_i))$$
  Où :
  - $\epsilon_t$ est l'erreur de classification pondérée à l'étape $t$.
  - $w_i^{(t)}$ est l'**hyperparamètre/paramètre dynamique** représentant le poids de la donnée $i$ à l'étape $t$.
  - L'**hyperparamètre** principal de l'ensemble est `n_estimators` (nombre maximal d'estimateurs faibles).
- **Avantages :** Très facile à implémenter ; n'exige pas de réglages complexes d'hyperparamètres pour donner de bons résultats.
- **Limites :** Extrêmement sensible au bruit et aux valeurs aberrantes (outliers), car il s'obstinera à sur-pondérer ces anomalies ; moins performant sur de très grandes dimensions sans sélection préalable.

## 17. K-means

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** K-moyennes (K-means)
- **Famille :** Partitionnement / Clustering (Apprentissage non supervisé)
- **Objectif :** Regrouper un ensemble d'observations non étiquetées en $K$ grappes (clusters) homogènes et distinctes sur la base de leur similarité géométrique.
- **Intuition et Étapes :**
  1. *Initialisation :* Sélection de $K$ points de l'espace comme centroïdes initiaux (souvent de manière optimisée via la méthode K-means++).
  2. *Affectation :* Chaque observation est assignée au centroïde dont elle est la plus proche (au sens d'une métrique, généralement la distance euclidienne).
  3. *Mise à jour :* La position de chaque centroïde est recalculée comme étant le centre de gravité (moyenne de coordonnées) des points qui lui ont été assignés.
  4. *Convergence :* Répétition des étapes 2 et 3 jusqu'à ce que les positions des centroïdes se stabilisent ou que le nombre maximal d'itérations soit atteint.
- **Formule mathématique essentielle :**
  L'algorithme cherche à minimiser l'inertie intra-classe (somme des carrés des écarts à la moyenne ou *Within-Cluster Sum of Squares*, WCSS) :
  $$\mathcal{J} = \sum_{k=1}^{K} \sum_{X_i \in S_k} \|X_i - \mu_k\|^2$$
  Où :
  - $S_k$ représente l'ensemble des observations affectées au cluster $k$.
  - $\mu_k$ est le centroïde du cluster $k$ (moyenne des points de $S_k$).
  - Le nombre de clusters $K$ (`n_clusters`) est l'**hyperparamètre** majeur de l'algorithme.
- **Avantages :** Très simple à comprendre, extrêmement rapide à calculer et performant sur des jeux de données de grande taille (complexité linéaire par rapport au nombre de données).
- **Limites :** Sensible au choix initial des centroïdes (peut converger vers un minimum local) ; ne performe bien que pour des clusters de formes sphériques et de tailles comparables ; nécessite de spécifier le nombre $K$ à l'avance.

### Présentation
- **Nom :** XGBoost (eXtreme Gradient Boosting / XGBClassifier)
- **Famille :** Classification / Régression (Ensemble Learning / Boosting - Apprentissage supervisé)
- **Objectif :** Optimiser l'algorithme classique de Gradient Boosting pour offrir une vitesse d'exécution extrême et des performances prédictives maximales en intégrant des pénalités de régularisation directement dans la fonction objective.
- **Intuition et Étapes :**
  1. *Entrée :* Variables explicatives $X$ et étiquettes de classes discrètes $y$.
  2. *Initialisation :* Calcul d'une prédiction de base constante (par exemple, 0.5 pour la probabilité logistique).
  3. *Optimisation de Taylor :* À chaque itération, l'algorithme utilise un développement de Taylor de second ordre pour approximer la fonction de perte globale et trouver la structure d'arbre optimale.
  4. *Régularisation :* Élagage des arbres à l'aide de pénalités sur le nombre de feuilles et la magnitude des poids de sortie.
  5. *Sortie :* Combinaison additive d'arbres prédictifs hautement optimisés.
- **Formule mathématique essentielle :**
  La fonction objective régularisée à l'étape $t$ à minimiser est :
  $$\mathcal{L}^{(t)} \approx \sum_{i=1}^m [g_i f_t(X_i) + \frac{1}{2} h_i f_t^2(X_i)] + \gamma T + \frac{1}{2} \lambda \sum_{j=1}^T w_j^2$$
  Où :
  - $g_i$ et $h_i$ sont respectivement la dérivée première (gradient) et seconde (hessienne) de la fonction de perte par rapport à la prédiction de l'étape précédente.
  - $T$ est le nombre de feuilles de l'arbre $f_t$ et $w_j$ représente le score attribué à la feuille $j$ (paramètres).
  - $\gamma$ (`gamma`) et $\lambda$ (`reg_lambda`) sont les **hyperparamètres** de régularisation qui contrôlent la complexité du modèle.
- **Avantages :** Vitesse de calcul exceptionnelle grâce à la parallélisation et à l'optimisation du cache ; gère automatiquement et efficacement les données manquantes.
- **Limites :** Un très grand nombre d'hyperparamètres à régler (nécessite une phase de recherche par grille ou bayésienne) ; risque important de surapprentissage si la régularisation n'est pas calibrée.

## 18. Classification hiérarchique

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** Classification hiérarchique ascendante (Hierarchical Agglomerative Clustering - HAC)
- **Famille :** Partitionnement / Clustering (Apprentissage non supervisé)
- **Objectif :** Regrouper des observations dans une structure arborescente de partitions imbriquées (dendrogramme) sur la base de leur distance mutuelle.
- **Intuition et Étapes :**
  1. *Initialisation :* Chaque point individuel commence par former son propre cluster unique ($N$ clusters au départ).
  2. *Regroupement itératif :* À chaque étape, l'algorithme identifie et fusionne les deux clusters les plus proches selon un critère de liaison (linkage).
  3. *Arrêt :* L'arborescence est complétée lorsqu'il ne reste plus qu'un seul grand cluster englobant toutes les observations.
  4. *Coupe :* L'utilisateur effectue une coupe horizontale du dendrogramme à un certain niveau de hauteur pour obtenir le nombre $K$ de clusters souhaité.
- **Formule mathématique essentielle :**
  La proximité entre deux clusters $A$ et $B$ dépend du critère de liaison choisi. Pour le critère de **liaison de Ward** (le plus classique), l'algorithme fusionne les groupes qui minimisent l'augmentation de la variance intra-classe totale :
  $$\Delta I_W(A, B) = \frac{|A| \cdot |B|}{|A| + |B|} \| \mu_A - \mu_B \|^2$$
  Où :
  - $|A|$ et $|B|$ désignent les effectifs respectifs des clusters $A$ et $B$.
  - $\mu_A$ et $\mu_B$ sont les vecteurs de coordonnées moyennes (centroïdes) de chaque groupe.
  - L'**hyperparamètre** principal est le type de lien (`linkage` : 'ward', 'complete', 'average', 'single') et la métrique de distance (`metric` : euclidienne, manhattan, etc.).
- **Avantages :** Produit un dendrogramme hautement informatif qui révèle l'organisation à différentes échelles ; n'impose pas d'initialisation aléatoire (les résultats sont strictement déterministes).
- **Limites :** Complexité temporelle et spatiale élevée (généralement de l'ordre de $\mathcal{O}(N^3)$ en calcul et $\mathcal{O}(N^2)$ en mémoire), rendant l'algorithme inutilisable sur de très grands jeux de données ; sensible aux valeurs aberrantes.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage

# Calcul de la matrice des liens hiérarchiques avec la méthode de Ward
Z_hier = linkage(X_clu, method='ward')

plt.figure(figsize=(10, 6))
# Tracé du dendrogramme (limité aux 30 dernières fusions pour la lisibilité)
dendrogram(Z_hier, truncate_mode='lastp', p=30, leaf_rotation=45, leaf_font_size=10, show_contracted=True)

# Ligne de coupe horizontale pour obtenir 3 groupes
plt.axhline(y=15.0, color='red', linestyle='--', linewidth=2, label='Seuil de coupe (K = 3)')

plt.title("Dendrogramme de la Classification Hiérarchique Ascendante")
plt.xlabel("Index ou effectif des groupes d'échantillons")
plt.ylabel("Distance de liaison (Inertie de Ward)")
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

- **Cas peu approprié :** Analyse transactionnelle de masse en temps réel (ex: des millions de paiements d'utilisateurs par jour), à cause de la complexité spatiale $\mathcal{O}(N^2)$ requise pour stocker la matrice de distances en mémoire.
- **Métriques adaptées :** Coefficient de Silhouette, inertie intra-classe, et coefficient de corrélation cophénétique (évalue la fidélité de l'arbre par rapport aux distances d'origine).

## 19. DBSCAN

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** DBSCAN (Density-Based Spatial Clustering of Applications with Noise)
- **Famille :** Partitionnement / Clustering basé sur la densité (Apprentissage non supervisé)
- **Objectif :** Regrouper des observations en amas de formes arbitraires sur la base de la densité locale de points, tout en identifiant automatiquement les données atypiques comme du bruit.
- **Intuition et Étapes :**
  1. *Entrée :* Un ensemble d'observations non étiquetées.
  2. *Classification des points :* Pour chaque point, l'algorithme compte le nombre de voisins dans un rayon $\epsilon$ (Epsilon).
     - **Point cœur (Core point) :** s'il possède au moins $MinPts$ voisins dans son rayon $\epsilon$.
     - **Point frontière (Border point) :** s'il a moins de $MinPts$ voisins mais est situé dans le rayon $\epsilon$ d'un point cœur.
     - **Bruit (Noise) :** s'il n'est ni un point cœur, ni un point frontière.
  3. *Propagation et fusion :* Un cluster est formé par un point cœur et tous les points (cœurs ou frontières) qui lui sont accessibles par densité.
- **Formule mathématique essentielle :**
  Le voisinage d'un point $p$, noté $N_\epsilon(p)$, est défini par :
  $$N_\epsilon(p) = \{q \in X \mid \text{dist}(p, q) \le \epsilon\}$$
  La condition pour que $p$ soit un point cœur est :
  $$|N_\epsilon(p)| \ge MinPts$$
  Où les **hyperparamètres** fondamentaux sont $\epsilon$ (`eps` : distance maximale de voisinage) et $MinPts$ (`min_samples` : nombre minimal de points requis pour former une région dense).
- **Avantages :** N'exige pas de spécifier le nombre de clusters $K$ à l'avance ; capable de découvrir des clusters de formes géométriques complexes et arbitraires tout en ignorant le bruit.
- **Limites :** Très sensible au choix des hyperparamètres $\epsilon$ et $MinPts$ ; peine à identifier des structures de densités variables.

- **Cas peu approprié :** Jeux de données à haute dimensionnalité ou avec des densités extrêmement hétérogènes (les grappes de faible densité seront toutes classées comme du bruit tandis que les denses seront fusionnées).

### Présentation
- **Nom :** Modèles de mélanges gaussiens (Gaussian Mixture Models — GMM)
- **Famille :** Partitionnement / Clustering probabiliste (Apprentissage non supervisé)
- **Objectif :** Regrouper des observations en modélisant la distribution des données comme une superposition (mélange) de plusieurs distributions normales (gaussiennes) ayant chacune leurs propres paramètres de moyenne et de covariance.
- **Intuition et Étapes :**
  1. *Initialisation :* Choix du nombre de composantes $K$ (souvent initialisées via K-means pour les centres d'origine).
  2. *Étape E (Expectation) :* Calculer pour chaque point de donnée la probabilité (responsabilité) qu'il appartienne à chacune des $K$ composantes gaussiennes.
  3. *Étape M (Maximization) :* Mettre à jour les paramètres de chaque gaussienne (moyenne, matrice de covariance et poids de mélange) en maximisant la vraisemblance pondérée par les responsabilités calculées à l'étape E.
  4. *Convergence :* Répéter les étapes E et M jusqu'à stabilisation de la log-vraisemblance globale.
- **Formule mathématique essentielle :**
  La fonction de densité de probabilité du mélange pour une observation $x$ s'écrit :
  $$p(x) = \sum_{k=1}^{K} \pi_k \mathcal{N}(x | \mu_k, \Sigma_k)$$
  Où :
  - $\pi_k$ est le poids de mélange de la composante $k$, avec $\sum_{k=1}^K \pi_k = 1$ et $\pi_k \ge 0$ (paramètre).
  - $\mathcal{N}(x | \mu_k, \Sigma_k)$ est la densité gaussienne multivariée définie par sa moyenne $\mu_k$ et sa matrice de covariance $\Sigma_k$ (paramètres).
  - Le nombre de composantes $K$ (`n_components`) et le type de covariance (`covariance_type`) sont les principaux **hyperparamètres**.
- **Avantages :** Attribution probabiliste souple (*soft clustering*) permettant de quantifier l'incertitude d'affectation ; s'adapte à des clusters de tailles et d'orientations elliptiques très variées grâce aux matrices de covariance.
- **Limites :** Sensible aux minima locaux lors de l'optimisation par l'algorithme EM ; l'estimation des matrices de covariance complètes requiert un grand nombre de paramètres, créant des risques d'instabilité numérique si les données manquent.

### Schéma explicatif

Le code ci-dessous produit une visualisation des ellipses de confiance à 1 et 2 écarts-types pour chaque composante gaussienne ajustée par l'algorithme GMM, illustrant le partitionnement probabiliste et l'orientation des formes trouvées.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture
from matplotlib.patches import Ellipse

def tracer_ellipse(mu, cov, ax, n_std=2.0, **kwargs):
    """Trace une ellipse représentant la covariance d'une composante gaussienne."""
    valeurs_propres, vecteurs_propres = np.linalg.eigh(cov)
    ordre = valeurs_propres.argsort()[::-1]
    valeurs_propres, vecteurs_propres = valeurs_propres[ordre], vecteurs_propres[:, ordre]
    angle = np.degrees(np.arctan2(vecteurs_propres[1, 0], vecteurs_propres[0, 0]))
    largeur, hauteur = 2 * n_std * np.sqrt(valeurs_propres)
    ell = Ellipse(xy=mu, width=largeur, height=hauteur, angle=angle, **kwargs)
    ax.add_patch(ell)

# Ajustement GMM
gmm = GaussianMixture(n_components=3, random_state=42)
gmm.fit(X_clu)
etiquettes_gmm = gmm.predict(X_clu)

fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(X_clu[:, 0], X_clu[:, 1], c=etiquettes_gmm, cmap='viridis', s=30, alpha=0.7, zorder=2)

# Tracé des ellipses de dispersion pour chaque composante
couleurs = ['purple', 'teal', 'yellow']
for i in range(3):
    mu = gmm.means_[i]
    cov = gmm.covariances_[i]
    tracer_ellipse(mu, cov, ax, n_std=1.0, edgecolor='red', fill=False, linestyle='--', linewidth=1.5, label='1 Std Dev' if i==0 else "")
    tracer_ellipse(mu, cov, ax, n_std=2.0, edgecolor='black', fill=False, linestyle='-', linewidth=1.5, label='2 Std Dev' if i==0 else "")

plt.title("Modélisation probabiliste par mélanges de gaussiennes (GMM)")
plt.xlabel("Caractéristique X1")
plt.ylabel("Caractéristique X2")
plt.legend(loc='upper right')
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()
```

### Présentation
- **Nom :** Analyse en Composantes Principales (Principal Component Analysis - PCA)
- **Famille :** Réduction de dimension / Apprentissage non supervisé
- **Objectif :** Réduire la dimensionnalité d'un jeu de données contenant un grand nombre de variables corrélées, tout en conservant le maximum de la variance (l'information) d'origine.
- **Intuition et Étapes :**
  1. *Centrage et réduction :* Standardisation des données d'origine pour que chaque variable ait une moyenne nulle et une variance unitaire.
  2. *Matrice de covariance :* Calcul de la matrice de corrélation ou de covariance des variables standardisées.
  3. *Décomposition :* Recherche des valeurs propres et des vecteurs propres de cette matrice (via la décomposition en valeurs singulières - SVD).
  4. *Projection :* Sélection des $k$ premiers vecteurs propres (composantes principales) associés aux plus grandes valeurs propres et projection des données initiales sur ce nouvel espace orthogonal de dimension réduite.
- **Formule mathématique essentielle :**
  La projection d'une matrice de données standardisées $X$ sur le nouvel espace de dimension réduite s'écrit :
  $$T = X W$$
  Où :
  - $X$ est la matrice des données standardisées ($m \times n$).
  - $W$ est la matrice de passage ($n \times k$) dont les colonnes sont les $k$ vecteurs propres (chargements / *loadings*) de la matrice de covariance $X^T X$.
  - $T$ est la matrice des coordonnées des individus dans le nouvel espace des composantes principales ($m \times k$).
  - L'**hyperparamètre** essentiel est `n_components` (le nombre $k$ de composantes principales à conserver).
- **Avantages :** Permet de visualiser de grands ensembles de données complexes en 2D ou 3D et élimine la colinéarité entre les caractéristiques pour les modèles de régression ultérieurs.
- **Limites :** Les nouvelles variables projetées (les composantes principales) sont des combinaisons linéaires des variables d'origine et perdent leur interprétabilité physique immédiate ; sensible aux valeurs aberrantes.

## 22. t-SNE

Construire le modèle, afficher une métrique ou une sortie et interpréter ses limites. Les exemples n’utilisent pas de test final pour le réglage d’hyperparamètres.

### Présentation
- **Nom :** t-SNE (t-Distributed Stochastic Neighbor Embedding / Plongement de voisins stochastiques distribués en t)
- **Famille :** Réduction de dimension / Collecte de structures (Apprentissage non supervisé non linéaire)
- **Objectif :** Représenter un ensemble de données de haute dimensionnalité dans un espace de faible dimension (généralement 2D ou 3D) en préservant au maximum les relations de proximité locales (voisinages).
- **Intuition et Étapes :**
  1. *Mesure de similarité en haute dimension :* Pour chaque paire de points, on calcule une probabilité conditionnelle sous une loi normale que le premier choisisse le second comme voisin.
  2. *Mesure de similarité en basse dimension :* Dans l'espace projeté, on utilise une distribution de Student (loi t) à un degré de liberté pour mesurer les similarités de manière à éviter l'entassement des points (*crowding problem*).
  3. *Optimisation (Gradient Descent) :* On minimise la divergence de Kullback-Leibler (KL) entre les deux distributions de probabilités à l'aide d'une descente de gradient pour faire correspondre les voisinages.
- **Formule mathématique essentielle :**
  La divergence de Kullback-Leibler minimisée par t-SNE s'écrit :
  $$\mathcal{KL}(P || Q) = \sum_{i} \sum_{j} p_{j|i} \log \frac{p_{j|i}}{q_{j|i}}$$
  Où :
  - $p_{j|i}$ est la probabilité de transition mesurant la similarité entre les points de départ de haute dimension.
  - $q_{j|i}$ est la probabilité mesurant la similarité dans l'espace projeté de basse dimension (loi t).
  - L'**hyperparamètre** majeur est la **perplexité** (`perplexity`), qui règle grossièrement le nombre effectif de voisins que chaque point prend en compte lors de l'apprentissage.
- **Avantages :** Révèle de manière spectaculaire des structures de groupes complexes et non linéaires (courbures, variétés entrelacées) intraduisibles par l'ACP.
- **Limites :** Ne conserve pas les distances globales (l'espace entre deux clusters éloignés n'a aucune signification réelle) ; coût computationnel très élevé et résultats instables variant selon l'initialisation aléatoire.
